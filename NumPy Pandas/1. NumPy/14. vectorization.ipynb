# VECTORIZATION

### Vectorization means replacing explicit loops with fast, built in array operations. This leads to cleanner code and huge performance gains.

THE PROBLEM 

```python
# Create two lines
list1 = list(range(1000))
list2 = list(range(1000))

# Multiply using a for loop (element by element)
def for_loop_multiply_list(list1, list2):
    product_list = []
    for i in range(len(list1)):
        product_list.append(list1[i] * list2[i])
    return product_list
```

This method works, but Python loops are slow because they process one elemnet at a time.

THE SOLUTION 

```python
import numpy as np

array1 = np.array(list1)
array2 = np.array(list2)

# Multiply using NumPy (vectorized)
result = array1 * array2    # element-wise multiplication
```
