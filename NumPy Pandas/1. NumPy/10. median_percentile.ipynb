# Median and Percentiles

### Two simple ways to understand the center and spread of data

```python
#   Median - The Middle Value
#   Percentie - A Percentage cut-off
#   Median is the 50th percentile
```

```python
import numpy as np
sales = np.array([120, 150, 180, 200, 210, 220, 270])
print(sales)
```

#### Median

```python
median = np.median(sales)
print(f'Median: {median}')
```

#### Percentile

```python
percentile = np.percentile(sales, 50)
print(f'Percentile: {percentile}')
```
