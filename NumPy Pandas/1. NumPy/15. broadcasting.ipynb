# Broadcasting

### How NumPy applies one values across many elements automatically  

```python
# A smaller values can be applied to a larger array.
# NumPy matches shapes automatically when possible.
# This lets you write less code.
```

TIP - Broadcasting is the engine behind vectorized operations. NumPy performs the work in optimized low-level code, so your programs run faster and your code stays clean and simple.

```python
import numpy as np

test_array = np.array([[1,2,3],
                       [1,2,3],
                       [1,2,3]])

result = test_array + 1
result 
```

```python
test_array[0, :] + test_array[:, 1].reshape(3,1)
```
