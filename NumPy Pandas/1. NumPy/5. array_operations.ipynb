```python
import numpy as np
arr_a = np.array([10,20,30,40,50])
arr_b = np.array([1,2,3,4, 5])
```

### Add a scalar

Adds 5 to every element 

```python
arr_a + 5
```

### Subtract a scalar

Subtracts 5 from every element

```python
arr_a - 5
```

### Multiply by a scalar

Multiples every element by 2

```python
arr_a * 2
```

### Divide by a scalar

Divides every element by 10

```python
arr_a / 10
```

### Element-wise addition

Adds matching positions

```python
arr_a + arr_b
```

### Elements-wise multiplication

Multiplies matching positions

```python
arr_a * arr_b
```
