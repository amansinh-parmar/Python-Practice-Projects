## 1D Array

```python
import numpy as np 
marks = np.array([72, 45, 88, 61, 95, 53])
```

```python
marks.sort()        # sort original array
print(marks)
sort_mark = np.sort(marks)     # create new variable and sort in new sorted array
print(sort_mark)
```

## 2D Array Matrix

```python
sales_2d = np.array([
    [52, 18, 41, 77, 29, 63],
    [34, 91, 25, 48, 56, 12],
    [83, 39, 67, 21, 45, 30],
    [14, 72, 58, 36, 90, 27]
])
```

Row-wise Sorting

```python
sales_2d.sort(axis=1)
print(f'Row wise Sort: {sales_2d}')
```

Column-wise Sorting

```python
sales_2d.sort(axis=0)
print(f'Column wise Sort: {sales_2d}')
```
