## A teacher has exam scores for 8 students and wants to analyze them with NumPy. Complete the Following Task:

Create a 2 X 4 NumPy array called score using these values:
35, 42, 50, 61, 28, 74, 81, 39

```python
import numpy as np
scores = np.array([35, 42, 50, 61, 28, 74, 81, 39])
scores = scores.reshape(2,4)
print(scores)
```

Add 5 grace marks to every score and store the result in update_score

```python
update_scores = scores + 5
print(f'Update Score: {update_scores}')
```

From update_scores, extract only the second row

```python
print(f'Second Row: {scores[1, :]}')
```

Filter the values in update_socres that are 50 or higher

```python
values = (update_scores[update_scores >= 50])
print(f'Score 50 or Higher: {values}')
```

Use np.where() to create a new array that shows "Pass" for scores 50 or more and "Needs Work" for scores below 50

```python
result = np.where(update_scores >= 50, 'Pass', 'Needs Work')
print(f'Result: {result}')
```
