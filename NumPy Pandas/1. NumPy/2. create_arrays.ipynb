```python
import numpy as np
```

* ones (create an array filled with ones)

- Syntax -  np.ones((rows, cols), dtype)

```python
np.ones((1, 4))
```

* zeros (create an array filled with zeros)

- Syntax - np.zeros((rows, cols), dtype)

```python
np.zeros((2,7), dtype=int)
```

* arange (create evenly stepped values in a range)

- Syntax - np.arange(start, stop, step)

```python
np.arange(10, 21, 2)
```

- linspace (creates evenly spanced values between two points)

* Syntax - np.linspace(start, stop, n)

```python
np.linspace(1, 5, 7, dtype=int)
```

* reshape (changes an array into a new compatible shape)

* Syntax - array.reshape(row, cols)

```python
np.arange(1, 13).reshape(3, 4)
```
