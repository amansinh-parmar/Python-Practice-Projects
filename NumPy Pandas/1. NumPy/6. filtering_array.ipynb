## Filter Arrays with Multiple Conditions

## Useful Pattern - mask = condition , filtered = ages[mask]

```python
import numpy as np
ages = np.array([14, 9, 52, 18, 28, 78, 42, 16])
```

### Only Adults

```python
adult_age = ages[ages >= 18]
print(f'Adult Age: {[adult_age]}')
```

### Only Children

```python
child_age = ages[ages < 18]
print(f'Child Age: {child_age}')
```

## Logical Rules 

#### & mean AND - both conditions must be true

#### | mean OR - either conditions can be true 

#### () Use parentheses around each condition

### Only Teenager 

```python
teen_age = (ages >= 15) & (ages <= 18)
print(f'Teenager Age: {ages[teen_age]}')
```

### Only Working Adults

```python
working_age = (ages >= 18) & (ages <= 60)
print(f'Wroking Age: {ages[working_age]}')
```

### Only Very Young and Old Ages 

```python
special_mask = (ages < 10) | (ages > 60)
print(f'Special Age: {ages[special_mask]}')
```

## Useful Example: 

```python
ages = np.array([12, 15, 17, 34, 45, 8, 19, 62, 29, 15, 51, 38, 70, 22, 16, 41])
names = np.array([
    "Aarav", "Priya", "Ananya", "Vikram", "Isha", "Arjun", "Meera", "Karan", "Diya", "Rahul", "Sneha", "Rajesh", "Kavya", "Aaditya", "Pooja", "Anushka"
])
```

# Modifying Array Values

```python
scores = np.array([45, 67, 32, 59, 88, 37])
print(scores)
```

### Change one value

```python
scores[0] = 50
print(scores)
```

### Replace a slice

```python
scores[1:3] = [70, 75]
print(scores)
```

### Update with condition

```python
scores[scores < 40] = 40
print(scores)
```

### Update selected positions

```python
scores[[3, 4]] = [60, 90]
print(scores)
```
