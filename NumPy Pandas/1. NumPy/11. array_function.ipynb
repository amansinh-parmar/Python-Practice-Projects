# ARRAY FUNCTION

Some NumPy function help you extract unique values or apply mathematical operations element-wise

```python
import numpy as np
sales = np.array([150, 200, 150, 300, 250, 200, 350, 150])
print(sales)
```

 Note - Use unique() to remove duplicates and sqrt() to transform values by taking their square root

#### Unique Values - Get the distinct (non-duplicate) values from the array

```python
unique = np.unique(sales)
print(f'Unique Values: {unique}')
```

#### Square Root (element-wise) - Compute the square root of each value in the array

```python
square_root = np.sqrt(sales)
print(f'Square Root: {square_root}')
```

## MORE USEFUL FUNCTIONS

```python
store_sales = np.array([120, 200, 150, 180, 210, 190, 170])
```

```python
#   Standard Deviation: How speard out the values are
store_sales.std()
```

```python
#   Position (index) of the highest and lowest sale
print(store_sales.argmax(), store_sales.argmin())
```

```python
#   Running Total: Each values = sum of all value up to that point
store_sales.cumsum()
```

```python
#   Sort values from smallest to largest 
np.sort(store_sales)
```

```python
#   Round each value to 2 decimal places
np.round(np.sqrt(store_sales), 2)
```

```python
#   Absolute value: removes the minus sign (e.g. profit/loss -> size of change)
changes = np.array([-20, 15, -5, 30, -10])
np.abs(changes)
```
