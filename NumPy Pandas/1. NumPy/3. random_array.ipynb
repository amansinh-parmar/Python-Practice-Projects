* TIP - rng is a common variable name for the generator created by default_rng 

```python
import numpy as np
```

* default_rng (Create a random number genarator. Use a seed when you want reproducible results)

- SYNTAX - np.random.default_rng(seed)

```python
from numpy.random import default_rng
rng = default_rng(42)
```

* random (Generates random values from uniform distribution between 0 and 1)

- SYNTAX - rng.random(size)

```python
values = rng.random(5)
values
```

- EXERCISE - Find the 5 random numbers between 1 to 10 

```python
numbers = np.random.randint(1, 11, size=5)
print(f'Random numbers - {numbers}')


# Find unique number
num = np.random.choice(np.arange(1, 11), size=5, replace=False)
print(f'Unique random numbers - {num}')
```

* normal (Generates random values from a normal distribution using a mean and standard deviation)

- SYNTAX - rng.normal(loc, scale)

```python
scores = rng.normal(loc=70, scale=10, size=10)
scores
```

* - Quick Idea - default_rng makes the generator, random gives 0 to 1 decimals and normal gives bell-curve style values
