# Import NumPy

```python
import numpy as np
```

## Get the Array

```python
temprature = [54, 87, 92, 50, 42]
temp_array = np.array(temprature)
temp_array
```

### Get the Array 3D 'ndim' (How many Axes in Array)

### Shape 'shape' (How many axis in each elements in the Array)

### Size 'size' (Total number of values)

### Dtype 'dtype' (The type of data stored) 

```python
print(temp_array.ndim)
print(temp_array.shape)
print(temp_array.size)
print(temp_array.dtype)
```

## Get the 2D Array

```python
temprature = [[1,2,3], [4,5,6]]
temp_array = np.ndim(temprature)
temp_array
```
