# Mixed NumPy Practice

## A teacher has 8 quiz scores and wants to practice NumPy on them. Complete the following tasks.

Create a 1D NumPy array named "scores" using: 16, 25, 36, 25, 49, 64, 81, 100

```python
import numpy as np 
scores = np.array([16, 25, 36, 25, 49, 64, 81, 100])
```

Update every score below 25 to 25


```python
scores[scores < 25] = 25
print(scores)
```

Create a new array named "sorted_scores" by sorting the updated scores in ascending order

```python
sorted_scores = np.sort(scores)
print(f'Sorted Scores: {sorted_scores}')
```

From the updated array, find the total, average, highest value and median

```python
# Total
total = np.sum(sorted_scores)
print(f'Total Score: {total}')
# Average
avg = np.mean(sorted_scores)
print(f'Average Score: {avg}')
# Highest 
highest = sorted_scores.max()
print(f'Highest Score: {highest}')
# Median
middle = np.median(sorted_scores)
print(f'Middle or Median Score: {middle}')
```

Use "np.unique()" to get the distinct score value

```python
unique = np.unique(sorted_scores)
print(f'Unique Score: {unique}')
```

Use "np.sqrt()" to calculate the square root of each updated score

```python
square_root = np.sqrt(sorted_scores)
print(f'Square Root: {square_root}')
```
