# THE WHERE FUNCTION

### SYNTAX - np.where(condition, value_if_true, value_if_false)

- Convert grades: pass if >= 50 else fail

```python
import numpy as np
scores = np.array([45, 32, 67, 90, 58])
result = np.where(scores >= 50, "Pass", "Fail")
print(f'Result: {result}')
```

- Make numbers positive (replace negatives with 0)

```python
nums = np.array([-4, -1, 3, -8, 7, 0])
positive_num = np.where(nums <= 0, 0, nums)
print(f'Positive Numbers: {positive_num}')
```

- Apply discount: 10% off if price > 100

```python
prices = np.array([80, 120, 150, 90, 200])
discount_price = np.where(prices > 100, prices * 0.9, prices)
print(f'Discount Price: {discount_price}')
```

- Categorize ages

```python
ages = np.array([8, 15, 22, 35, 60, 70])
category = np.where(ages > 18, 'Adult', 'Child')
print(f'Age Category: {category}')
```

- Replace even numbers with 'Even' and odd with 'Odd'  

```python
numbers = np.array([1,2,3,4,5,6,7])
odd_even = np.where(numbers % 2 == 0, 'Even', 'Odd')
print(f'Number Category: {odd_even}')
```

- Cap values at a maximum (Upper bound)

```python
values = np.array([10, 25, 50, 75, 120])
capped = np.where(values > 100, 100, values)
print(f'Cap Values: {capped}')
```

# Exersice:

### Q1. Weather report: Label each day "Hot" if temperature > 35, else "Normal"

```python
temperature = np.array([28, 36, 41, 33, 30, 38, 25])
report = np.where(temperature > 35, 'Hot', 'Normal')
print(f'Temprature Report: {report}')
```

### Q2. Absent student are marked -1: Replace -1 with 0 and keep other marks as they are

```python
marks = np.array([78, -1, 56, 89, -1, 45])
replace = np.where(marks == -1, 0, marks)
print(f'Marks: {replace}')
```

### Q3. Free delivery: Delivery charges is 0 if order amount >= 499, else 40

```python
orders = np.array([250, 499, 799, 120, 1500])
charges = np.where(orders >= 499, orders, orders + 40)
print(f'Delivery Charges with Price: {charges}')
```

### Q4. Salary Hike: 20% hike if salary < 3000, else 10% hike. Find the new salaries

```python
salaries = np.array([25000, 45000, 18000, 60000, 30000])
new_salary = np.where(salaries < 3000, salaries * 1.20, salaries * 1.10)
print(f'New Salary: {new_salary}')
```

### Q5. Phone battery status (use two arrays togather):

```python
# "Charging"      ->  if the phone is plugged in
# "Charge Now"    ->  if NOT plugged in AND battery < 20
# "OK"            ->  otherwise  
battery = np.array([15, 80, 5, 45, 19, 10])
is_charging = np.array([False, False, True, False, False, True])
status = np.where(is_charging == True, "Charging", 
                  np.where(battery < 20, "Charge Now", "OK"))
print(f'Battery Status: {status}')
```

### Q6. (Challenge) Stock report using yesterday's and today's prices:

```python
#       First find the % change = (today - yesterday) / yesterday * 100
#       Then label: "Big Jump" if change > 5, "UP" if change > 0,
#                   "Down" if change < 0, else "No Change"
#       Hint: put one np.where inside another

yesterday = np.array([200, 150, 320, 90, 500, 75])
today = np.array([215, 150, 310, 93, 500, 72])

changes = (today - yesterday) / yesterday * 100
label = np.where(changes > 5, "BIG JUMP", 
                 np.where(changes > 0, "UP",
                          np.where(changes < 0, "DOWN", "No Changes")))
print(f'Yesterday Price: {yesterday}\nToday Price: {today}')
print(f"Today's Stock Price Change: {label}")
```
