# Array Aggregation Method

### Use NumPy to summarize sales data quickly

```python
import numpy as np
sales = np.array([120, 200, 150, 180, 210, 190, 170])
print(sales)
```

 Total Sales    - Adds all sales values

```python
sum = sales.sum()
print(f'Sum: {sum}')
```

Average Sales

```python
average = sales.mean()
round = np.round(average, 2)
print(f'Average: {round}')
```

Highest Sales

```python
highest = sales.max()
print(f'Highest Sale: {highest}')
```

Lowest Sales

```python
lowest = sales.min()
print(f'Lowest Sale: {lowest}')
```

## Summarize 2D sales data across the whole array, rows or columns

```python
company_sales = np.array([[120, 150, 180, 200], 
                         [100, 170, 160, 210],
                         [140, 130, 190, 220]])
print(company_sales) 
```

Whole Array Total

```python
total = company_sales.sum()
print(total)
```

Column Wise Sum

```python
column = company_sales.sum(axis=0)
print(f'Column 0 Sum: {column}')
```

Row Wise Sum

```python
row = company_sales.sum(axis=1)
print(f'Row 1 Sum: {row}')
```

Column Wise Average

```python
col_avg = company_sales.mean(axis=0)
print(f'Column 0 Average: {col_avg}')
```

Row Wise Average

```python
row_avg = company_sales.mean(axis=1)
print(f'Row 1 Average: {row_avg}')
```

Highest Sales per Store

```python
high_sale_store = company_sales.max(axis=1)
print(f'Highest Sales per Store: {high_sale_store}')
```

Lowest Sales per Store

```python
low_sale_store = company_sales.min(axis=1)
print(f'Lowest Sales per Store: {low_sale_store}')
```
